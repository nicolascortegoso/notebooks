# Latent Semantic Analysis (LSA)

In LSA, a term-document matrix (where rows represent words and columns represent documents) is decomposed using SVD to identify latent semantic relationships, reducing noise and capturing underlying concepts (e.g., grouping related terms like "car" and "automobile") by focusing on the most significant singular values and vectors.

Import libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
```

A small collection of documents

```python
corpus = [
    "tacos is a dish from Mexico",
    "spaghetty is a dish from Italy",
    "football is a sport from Italy",
    "boxing is a sport from Mexico",
    "tacos are popular in Mexico",
    "pizza is a famous dish from Italy",
    "soccer is a sport from Mexico",
    "spaghetti is an Italian dish",
    "boxing is a popular sport in Mexico",
    "pasta is a famous Italian dish",
    "football is a sport from Italy",
    "burritos are from Mexico",
    "pizza is from Italy",
    "football is a famous sport from Italy",
    "salsa is a dance from Mexico",
    "tacos are widely known in Mexico",
    "soccer is a famous sport in Mexico",
    "spaghetti is a well-known Italian dish",
    "pasta is from Italy",
    "burritos are from Mexico",
    "Italian pizza is famous worldwide",
    "Mexican tacos are delicious",
    "Italian pasta dishes are delicious",
    "football is a sport from Italy",
    "boxing is a popular sport in Mexico"
]
```

Initializes a TF-IDF vectorizer to convert the text corpus into a document-term matrix. The `TfidfVectorizer` is configured to exclude English stop words (e.g., "the", "is") to focus on meaningful terms. The `fit_transform` method processes the corpus, generating a sparse document-term matrix where rows represent documents, columns represent terms, and values are TF-IDF scores.

```python
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(corpus)
```

Extracts the feature names (terms) from the TF-IDF vectorizer and converts the sparse document-term matrix into a pandas DataFrame for visualization.
Each row represents a document, each column represents a term, and the values are TF-IDF scores rounded to two decimal places.
Displaying the DataFrame provides a human-readable view of the TF-IDF features.

```python
feature_names = vectorizer.get_feature_names_out()

df_tfidf = pd.DataFrame(X.toarray().round(2), columns=feature_names)
df_tfidf
```

Applies Latent Semantic Analysis (LSA) using `TruncatedSVD` to reduce the document-term matrix to a 2-topic space. Projects documents into this space with `fit_transform`, producing coordinates representing relevance to each topic. Values closer to 1 or -1 indicate stronger topic association.

```python
n_topics = 2
lsa = TruncatedSVD(n_components=n_topics, random_state=42)
X_lsa = lsa.fit_transform(X)
```

Creates a pandas DataFrame to display coordinates, with columns `Topic 1` and `Topic 2`, and includes original document text. Sorts the DataFrame by `Topic 1` in descending order to highlight documents most associated with that topic.

```python
df_lsa = pd.DataFrame(X_lsa, columns=[f'Topic {i}' for i in range(1, n_topics + 1)])
df_lsa["Document"] = corpus
df_lsa.sort_values(f"Topic 1", ascending=False)
```

Inspects the top terms associated with each LSA topic to interpret their meaning.

```python
top_n_terms = 5

components_df = pd.DataFrame(lsa.components_, columns=feature_names, index=[f'Topic {i}' for i in range(1, n_topics + 1)])

for topic in range(1, n_topics + 1):
  display(components_df.T[f'Topic {topic}'].sort_values(ascending=False).head(top_n_terms))
```
